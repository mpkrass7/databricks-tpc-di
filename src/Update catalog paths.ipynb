
# Find and replace catalog path in notebooks

We're looking for ${wh_db}_${scale_factor} and replacing with ${wh_db}

```python
from pathlib import Path
import os
path_to_files = Path("single_batch/SQL/")
remove_text = "${wh_db}_${scale_factor}"
replace_text = "${wh_db}"

for file in os.listdir(path_to_files):
  file_path = f"file:{(path_to_files / file).absolute()}"
  if (path_to_files / file).is_file():
    print(file_path)
    with open(file_path, mode='r') as fin:
      lines = [line.replace(remove_text, replace_text) for line in fin]

    with open(file_path, "wt") as fout:
        for line in fin:
            fout.write(line.replace('${wh_db}_${scale_factor}', '${wh_db}'))
    break
```
